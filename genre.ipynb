```python
import pandas as pd

# Load dataset
df = pd.read_csv("movie_metadata.csv")

# Basic preprocessing
# Replace missing values
df = df.fillna("")

# Convert all text columns to lowercase for easy matching
text_cols = df.select_dtypes(include=['object']).columns
df[text_cols] = df[text_cols].apply(lambda col: col.str.lower())

# Split genres into lists
df["genres"] = df["genres"].apply(lambda x: x.split("|") if isinstance(x, str) else [])

# Combine actor columns into a single list
df["actors"] = df.apply(lambda row: [
        row["actor_1_name"], 
        row["actor_2_name"], 
        row["actor_3_name"]
    ], axis=1)

```

```python
def get_top10_movies(df, genre=None, actor=None):
    genre = genre.lower() if genre else None
    actor = actor.lower() if actor else None

    filtered = df

    # Filter by genre
    if genre:
        filtered = filtered[filtered["genres"].apply(lambda g: genre in g)]

    # Filter by actor
    if actor:
        filtered = filtered[filtered["actors"].apply(lambda a_list: actor in a_list)]

    # Sort by IMDb score (highest first)
    filtered = filtered.sort_values(by="imdb_score", ascending=False)

    # Take top 10 movie names
    top10 = filtered["movie_title"].head(10).reset_index(drop=True)

    # Change index to start from 1
    top10.index = top10.index + 1 

    return top10
```

```python
print(get_top10_movies(df, genre="action"))
```
