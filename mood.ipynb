```python
# Simple NLP Mood Detector + Movie Genre Mapper

import re

# 1. Mood keywords dictionary
mood_keywords = {
    "sad": ["sad", "depressed", "unhappy", "down", "low", "upset"],
    "happy": ["happy", "joy", "excited", "glad", "delighted"],
    "lonely": ["lonely", "alone", "isolated", "feeling left out"],
    "angry": ["angry", "mad", "furious", "annoyed", "irritated"],
    "romantic": ["love", "romantic", "affection", "crush", "date"],
    "bored": ["bored", "nothing to do", "tired", "dull"],
    "stressed": ["stress", "tense", "anxious", "worried"]
}

# 2. Mood → Movie genre mapping
mood_to_genre = {
    "sad": "Uplifting movies (inspirational, feel-good, emotional healing)",
    "happy": "Comedy movies",
    "lonely": "Feel-good, Friendship, or Heartwarming movies",
    "angry": "Action / Thriller movies",
    "romantic": "Romantic movies or romantic comedies",
    "bored": "Adventure / Fantasy movies",
    "stressed": "Calm, soothing genres like Slice-of-life or Light drama"
}

def detect_mood(sentence):
    sentence = sentence.lower()

    for mood, keywords in mood_keywords.items():
        for word in keywords:
            if re.search(r"\b" + word + r"\b", sentence):
                return mood

    return "neutral"

def recommend_genre(sentence):
    mood = detect_mood(sentence)

    if mood in mood_to_genre:
        return f"Mood detected: {mood}\nRecommended genre: {mood_to_genre[mood]}"
    else:
        return "Mood not detected. Try describing your feelings."

# ---------------------------
# Test the system
# ---------------------------

inputs = [
    "I am feeling sad today",
    "I am super happy!",
    "I am bored",
    "Feeling so much love",
    "I'm very stressed right now"
]

for text in inputs:
    print(text, "→", recommend_genre(text), "\n")
```
